# Unleased public API — statistical analysis

A5 Part 3.4 — Alternative use #3: pulling the public `listings/api/by-status/` endpoint into a notebook and running a simple statistical analysis on it (a confidence interval for a proportion), the same way a data analyst outside this project might.

```python
import math
import requests

PUBLIC_API_URL = "https://parulmudaliar.pythonanywhere.com/listings/api/by-status/"
rows = requests.get(PUBLIC_API_URL, timeout=10).json()
rows
```

Each row is one status category with its listing count — exactly the chart-ready shape the Vega-Lite bar chart also consumes from this same public endpoint.

```python
by_status = {row["status"]: row["total"] for row in rows}
by_status
```

## Confidence interval for a proportion

What fraction of listings are currently `Available`? Treat each listing as a coin flip (Available vs. not) and compute the 95% confidence interval for that proportion using the normal approximation to the binomial — the same idea behind a political poll's margin of error:

$$ SE = \sqrt{\frac{\hat p (1-\hat p)}{n}} \qquad CI_{95\%} = \hat p \pm 1.96 \times SE $$

```python
n = sum(by_status.values())
available = by_status.get("Available", 0)
p_hat = available / n if n else 0.0

se = math.sqrt(p_hat * (1 - p_hat) / n) if n else 0.0
z_95 = 1.96
margin = z_95 * se
ci_low, ci_high = max(0.0, p_hat - margin), min(1.0, p_hat + margin)

print(f"n = {n} listings")
print(f"p_hat (proportion Available) = {p_hat:.3f}")
print(f"standard error                = {se:.3f}")
print(f"95% confidence interval       = [{ci_low:.3f}, {ci_high:.3f}]")
```

**Interpretation:** if the current listing mix is a snapshot of a stable underlying process, the true share of `Available` listings would be expected to fall within **[44.9%, 100.0%]** 95% of the time. The interval is wide because `n=5` is a small sample — as Unleased gets more real listings, pulling this same public endpoint again would narrow it automatically, with no code changes needed on either side.
